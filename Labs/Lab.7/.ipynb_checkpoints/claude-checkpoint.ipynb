#### Part a - Pair plots
Using the compare_distributions and compare_scatter functions from Lecture 13a.

```python
# functions copied from lecture 13a
def compare_distributions(df, column_name, selections, **kwargs):
    for label, selection in selections.items():
        _=plt.hist(df[selection][column_name], label=label, **kwargs)
    _=plt.legend(fontsize=6)

def compare_scatter(df, x_var_name, y_var_name, selections, make_legend=True):
    for label, selection in selections.items():
        _=plt.scatter(df[selection][x_var_name], df[selection][y_var_name], label=label, s=1, alpha=0.3)
    _=plt.xlabel(x_var_name, fontsize=6)
    _=plt.ylabel(y_var_name, fontsize=6)
    if make_legend:
        _=plt.legend(fontsize=6)
```

```python
# pair plot for raw features - copied structure from lecture 13a pair plot
selection_dict = {"signal": df["signal"]==1, "background": df["signal"]==0}

# use a sample to speed things up
df_sample = df.sample(n=10000, random_state=42)
sel_sample = {"signal": df_sample["signal"]==1, "background": df_sample["signal"]==0}

columns = RawNames
n_columns = len(columns)
plt.figure(figsize=(18,18))

plot_i = 0
for i, x_var_name in enumerate(columns):
    for j, y_var_name in enumerate(columns):
        plot_i += 1
        plt.subplot(n_columns, n_columns, plot_i)
        make_legend = plot_i == 1
        if i == j:
            compare_distributions(df_sample, x_var_name, sel_sample,
                alpha=0.5, density=1, bins=30)
        else:
            compare_scatter(df_sample, x_var_name, y_var_name, sel_sample, make_legend=make_legend)

plt.suptitle("Pair Plot - Raw Features", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# pair plot for high-level features
columns = FeatureNames
n_columns = len(columns)
plt.figure(figsize=(18,18))

plot_i = 0
for i, x_var_name in enumerate(columns):
    for j, y_var_name in enumerate(columns):
        plot_i += 1
        plt.subplot(n_columns, n_columns, plot_i)
        make_legend = plot_i == 1
        if i == j:
            compare_distributions(df_sample, x_var_name, sel_sample,
                alpha=0.5, density=1, bins=30)
        else:
            compare_scatter(df_sample, x_var_name, y_var_name, sel_sample, make_legend=make_legend)

plt.suptitle("Pair Plot - High-level Features", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# part c - nice tables with tabulate
# part d - wrap it in a function

def show_matrix_table(matrix, names, title=""):
    # round to 2 decimal places
    table = list()
    for i in range(len(names)):
        row = [names[i]]
        for j in range(len(names)):
            row.append(round(matrix[i][j], 2))
        table.append(row)
    
    print(title)
    display(HTML(tabulate.tabulate(table, tablefmt='html', headers=[""] + names)))

def compute_and_show(df, var_names, title_prefix=""):
    corr = np.corrcoef(df[var_names].values, rowvar=False)
    cov = np.cov(df[var_names].values, rowvar=False)
    show_matrix_table(corr, var_names, title=title_prefix + " Correlation Matrix")
    show_matrix_table(cov, var_names, title=title_prefix + " Covariance Matrix")
    return corr, cov
```

## Exercise 5: Selection

### Exercise 5.1
Looking at the signal vs background distributions from earlier, for each observable I determine which cut type works best. For most of the high-level features, the signal tends to have higher values, so x > x_c works. For some variables the distributions overlap more symmetrically so |x - mu| type cuts could help.

```python
# Exercise 5.1 - determine best selection type for each variable
# by looking at the means of signal vs background
# if signal mean > background mean, use x > x_c (type 1)
# if signal mean < background mean, use x < x_c (type 2)

cut_types = dict()
for var in VarNames[1:]:
    sig_mean = np.mean(df_sig[var])
    bkg_mean = np.mean(df_bkg[var])
    if sig_mean > bkg_mean:
        cut_types[var] = "x > x_c"
    else:
        cut_types[var] = "x < x_c"
    print(var, ":", cut_types[var], " (sig mean:", round(sig_mean, 3), "bkg mean:", round(bkg_mean, 3), ")")
```

### Exercise 5.2
Plot TPR and FPR as function of x_c for each observable.

```python
# Exercise 5.2 - TPR and FPR as function of x_c
# TPR = fraction of signal that passes the cut
# FPR = fraction of background that passes the cut

def compute_tpr_fpr(df_sig, df_bkg, var, cut_values, cut_type):
    tpr = list()
    fpr = list()
    n_sig = len(df_sig)
    n_bkg = len(df_bkg)
    
    for xc in cut_values:
        if cut_type == "x > x_c":
            sig_pass = np.sum(df_sig[var] > xc)
            bkg_pass = np.sum(df_bkg[var] > xc)
        else:
            sig_pass = np.sum(df_sig[var] < xc)
            bkg_pass = np.sum(df_bkg[var] < xc)
        
        tpr.append(sig_pass / float(n_sig))
        fpr.append(bkg_pass / float(n_bkg))
    
    return np.array(tpr), np.array(fpr)
```

```python
# plot TPR and FPR for each variable
for var in VarNames[1:]:
    # make a range of cut values based on the data
    all_vals = df[var]
    cut_values = np.linspace(np.percentile(all_vals, 1), np.percentile(all_vals, 99), 100)
    
    cut_type = cut_types[var]
    tpr, fpr = compute_tpr_fpr(df_sig, df_bkg, var, cut_values, cut_type)
    
    plt.figure(figsize=(8, 4))
    plt.plot(cut_values, tpr, 'r-', label="TPR (signal eff)")
    plt.plot(cut_values, fpr, 'b-', label="FPR (bkg eff)")
    plt.xlabel("x_c")
    plt.ylabel("Efficiency")
    plt.title(var + " (" + cut_type + ")")
    plt.legend()
    plt.show()
```

### Exercise 5.3
Plot significance for each observable under 4 different scenarios.

```python
# Exercise 5.3 - significance plots
# sigma = N'_S / sqrt(N'_S + N'_B)
# where N'_S = eps_S * N_S, N'_B = eps_B * N_B

scenarios = [(10, 100), (100, 1000), (1000, 10000), (10000, 100000)]

def compute_significance(tpr, fpr, N_S, N_B):
    N_S_prime = tpr * N_S
    N_B_prime = fpr * N_B
    # avoid division by zero
    denom = np.sqrt(N_S_prime + N_B_prime)
    denom[denom == 0] = 1e-10
    return N_S_prime / denom
```

```python
# plot significance for each variable under each scenario
for var in VarNames[1:]:
    all_vals = df[var]
    cut_values = np.linspace(np.percentile(all_vals, 1), np.percentile(all_vals, 99), 100)
    
    cut_type = cut_types[var]
    tpr, fpr = compute_tpr_fpr(df_sig, df_bkg, var, cut_values, cut_type)
    
    plt.figure(figsize=(8, 4))
    for N_S, N_B in scenarios:
        sig = compute_significance(tpr, fpr, N_S, N_B)
        plt.plot(cut_values, sig, label="N_S=" + str(N_S) + " N_B=" + str(N_B))
    
    plt.xlabel("x_c")
    plt.ylabel("Significance")
    plt.title(var + " (" + cut_type + ")")
    plt.legend()
    plt.show()
```

## Exercise 6: Cut Flow

### Exercise 6.1
Choosing 3 observables that had the best separation based on the significance plots. Looking at the plots, M_R, M_TR_2, and R tend to have good separation.

```python
# Exercise 6.1 / 6.2 - cut flow
# pick the best observables and cut values from the significance plots
# i looked at the significance plots and picked cuts where significance peaks

# find optimal cut values by scanning
def find_best_cut(df_sig, df_bkg, var, cut_type, N_S, N_B):
    all_vals = df[var]
    cut_values = np.linspace(np.percentile(all_vals, 1), np.percentile(all_vals, 99), 200)
    tpr, fpr = compute_tpr_fpr(df_sig, df_bkg, var, cut_values, cut_type)
    sig = compute_significance(tpr, fpr, N_S, N_B)
    best_idx = np.argmax(sig)
    return cut_values[best_idx], sig[best_idx]

# pick 3 good variables
cut_vars = ["M_R", "M_TR_2", "R"]

print("Best cuts for scenario N_S=1000, N_B=10000:")
for var in cut_vars:
    best_xc, best_sig = find_best_cut(df_sig, df_bkg, var, cut_types[var], 1000, 10000)
    print(var, ": x_c =", round(best_xc, 3), "significance =", round(best_sig, 3))
```

```python
# Exercise 6.2 - cut flow table
# apply cuts one by one and track how signal and background efficiencies change

def cut_flow(df_sig, df_bkg, cut_vars, cut_types_dict, N_S, N_B):
    # first find the best cut for each variable
    cuts = list()
    for var in cut_vars:
        best_xc, _ = find_best_cut(df_sig, df_bkg, var, cut_types_dict[var], N_S, N_B)
        cuts.append(best_xc)
    
    # now apply cuts one by one
    sig_mask = np.ones(len(df_sig), dtype=bool)
    bkg_mask = np.ones(len(df_bkg), dtype=bool)
    
    n_sig_total = len(df_sig)
    n_bkg_total = len(df_bkg)
    
    table = list()
    
    for i, var in enumerate(cut_vars):
        xc = cuts[i]
        if cut_types_dict[var] == "x > x_c":
            sig_mask = sig_mask & (np.array(df_sig[var]) > xc)
            bkg_mask = bkg_mask & (np.array(df_bkg[var]) > xc)
        else:
            sig_mask = sig_mask & (np.array(df_sig[var]) < xc)
            bkg_mask = bkg_mask & (np.array(df_bkg[var]) < xc)
        
        eps_s = np.sum(sig_mask) / float(n_sig_total)
        eps_b = np.sum(bkg_mask) / float(n_bkg_total)
        N_S_prime = eps_s * N_S
        N_B_prime = eps_b * N_B
        if N_S_prime + N_B_prime > 0:
            significance = N_S_prime / np.sqrt(N_S_prime + N_B_prime)
        else:
            significance = 0.0
        
        table.append([var, round(xc, 3), round(eps_s, 4), round(eps_b, 4),
                      round(N_S_prime, 2), round(N_B_prime, 2), round(significance, 3)])
    
    return table
```

```python
# show cut flow tables for each scenario
from IPython.display import HTML, display
import tabulate

headers = ["Variable", "x_c", "eps_S", "eps_B", "N'_S", "N'_B", "Significance"]

for N_S, N_B in scenarios:
    print("Scenario: N_S =", N_S, ", N_B =", N_B)
    table = cut_flow(df_sig, df_bkg, cut_vars, cut_types, N_S, N_B)
    display(HTML(tabulate.tabulate(table, tablefmt='html', headers=headers)))
    print()
```

### Exercise 6.3
If two variables are correlated, cutting on one changes the distribution of the other. This means the optimal cut on the second variable depends on whether the first cut has been applied. Looking at the correlation matrix, some high-level features are correlated with each other. To handle this, we can optimize the second cut after applying the first.

```python
# Exercise 6.3 - optimize second cut after first cut is applied
# apply cut on M_R first, then find best cut for M_TR_2 on the remaining events

var1 = "M_R"
var2 = "M_TR_2"
N_S = 1000
N_B = 10000

# get best cut for var1
xc1, _ = find_best_cut(df_sig, df_bkg, var1, cut_types[var1], N_S, N_B)
print("Best cut for", var1, ":", round(xc1, 3))

# apply var1 cut to get reduced datasets
if cut_types[var1] == "x > x_c":
    df_sig_cut = df_sig[df_sig[var1] > xc1]
    df_bkg_cut = df_bkg[df_bkg[var1] > xc1]
else:
    df_sig_cut = df_sig[df_sig[var1] < xc1]
    df_bkg_cut = df_bkg[df_bkg[var1] < xc1]

# now find best cut for var2 on the reduced data
eps_s1 = len(df_sig_cut) / float(len(df_sig))
eps_b1 = len(df_bkg_cut) / float(len(df_bkg))
N_S_after = eps_s1 * N_S
N_B_after = eps_b1 * N_B

xc2_after, sig_after = find_best_cut(df_sig_cut, df_bkg_cut, var2, cut_types[var2], N_S_after, N_B_after)
print("Best cut for", var2, "after", var1, "cut:", round(xc2_after, 3))

# compare: find best cut for var2 without var1 cut
xc2_before, sig_before = find_best_cut(df_sig, df_bkg, var2, cut_types[var2], N_S, N_B)
print("Best cut for", var2, "without", var1, "cut:", round(xc2_before, 3))
print()
print("The optimal cut value changes when we account for the correlation between the variables.")

# now reverse the order
print()
print("--- Reversed order ---")
xc2_first, _ = find_best_cut(df_sig, df_bkg, var2, cut_types[var2], N_S, N_B)
if cut_types[var2] == "x > x_c":
    df_sig_cut2 = df_sig[df_sig[var2] > xc2_first]
    df_bkg_cut2 = df_bkg[df_bkg[var2] > xc2_first]
else:
    df_sig_cut2 = df_sig[df_sig[var2] < xc2_first]
    df_bkg_cut2 = df_bkg[df_bkg[var2] < xc2_first]

eps_s2 = len(df_sig_cut2) / float(len(df_sig))
eps_b2 = len(df_bkg_cut2) / float(len(df_bkg))
xc1_after, _ = find_best_cut(df_sig_cut2, df_bkg_cut2, var1, cut_types[var1], eps_s2*N_S, eps_b2*N_B)
print("Best cut for", var1, "after", var2, "cut:", round(xc1_after, 3))
```

## Exercise 7: ROC Curves

```python
# Exercise 7.1 - ROC curves for top 3 observables
# ROC curve plots TPR vs FPR
# area under curve (AUC) tells us how good the variable is at separating

def compute_roc(df_sig, df_bkg, var, cut_type, n_points=200):
    all_vals = df[var]
    cut_values = np.linspace(np.percentile(all_vals, 0.5), np.percentile(all_vals, 99.5), n_points)
    tpr, fpr = compute_tpr_fpr(df_sig, df_bkg, var, cut_values, cut_type)
    return fpr, tpr

def compute_auc(fpr, tpr):
    # sort by fpr
    sorted_idx = np.argsort(fpr)
    fpr_sorted = fpr[sorted_idx]
    tpr_sorted = tpr[sorted_idx]
    # trapezoidal rule
    auc = np.trapz(tpr_sorted, fpr_sorted)
    return abs(auc)
```

```python
# Exercise 7.1 - overlay ROC for top 3 variables
top_vars = ["M_R", "M_TR_2", "R"]

plt.figure(figsize=(8, 8))
for var in top_vars:
    fpr, tpr = compute_roc(df_sig, df_bkg, var, cut_types[var])
    auc = compute_auc(fpr, tpr)
    plt.plot(fpr, tpr, label=var + " (AUC=" + str(round(auc, 3)) + ")")

plt.plot([0, 1], [0, 1], 'k--', label="random")
plt.xlabel("FPR (background efficiency)")
plt.ylabel("TPR (signal efficiency)")
plt.title("ROC Curves")
plt.legend()
plt.show()
```

```python
# Exercise 7.2 - function to make ROC plots easily
def plot_roc(df_sig, df_bkg, var_list, cut_types_dict, title="ROC Curves"):
    plt.figure(figsize=(8, 8))
    for var in var_list:
        fpr, tpr = compute_roc(df_sig, df_bkg, var, cut_types_dict[var])
        auc = compute_auc(fpr, tpr)
        plt.plot(fpr, tpr, label=var + " (AUC=" + str(round(auc, 3)) + ")")
    
    plt.plot([0, 1], [0, 1], 'k--', label="random")
    plt.xlabel("FPR (background efficiency)")
    plt.ylabel("TPR (signal efficiency)")
    plt.title(title)
    plt.legend()
    plt.show()
```

```python
# Exercise 7.2 - test the function with all features
plot_roc(df_sig, df_bkg, FeatureNames, cut_types, "ROC - High-level Features")
```

```python
# Exercise 7.3 - ROC curves after successive selections
# apply cuts one at a time and show how the ROC changes for remaining variables

remaining_vars = ["M_TR_2", "R"]

# before any cut
print("Before any cuts:")
plot_roc(df_sig, df_bkg, remaining_vars, cut_types, "ROC before cuts")

# after M_R cut
xc_mr, _ = find_best_cut(df_sig, df_bkg, "M_R", cut_types["M_R"], 1000, 10000)
if cut_types["M_R"] == "x > x_c":
    df_sig_1 = df_sig[df_sig["M_R"] > xc_mr]
    df_bkg_1 = df_bkg[df_bkg["M_R"] > xc_mr]
else:
    df_sig_1 = df_sig[df_sig["M_R"] < xc_mr]
    df_bkg_1 = df_bkg[df_bkg["M_R"] < xc_mr]

print("After M_R cut:")
plot_roc(df_sig_1, df_bkg_1, remaining_vars, cut_types, "ROC after M_R cut")

# after M_R + M_TR_2 cut
xc_mtr, _ = find_best_cut(df_sig_1, df_bkg_1, "M_TR_2", cut_types["M_TR_2"], 1000, 10000)
if cut_types["M_TR_2"] == "x > x_c":
    df_sig_2 = df_sig_1[df_sig_1["M_TR_2"] > xc_mtr]
    df_bkg_2 = df_bkg_1[df_bkg_1["M_TR_2"] > xc_mtr]
else:
    df_sig_2 = df_sig_1[df_sig_1["M_TR_2"] < xc_mtr]
    df_bkg_2 = df_bkg_1[df_bkg_1["M_TR_2"] < xc_mtr]

print("After M_R + M_TR_2 cuts:")
plot_roc(df_sig_2, df_bkg_2, ["R"], cut_types, "ROC after M_R + M_TR_2 cuts")
```

```python
# Exercise 7.4 - change order of selections
# now do M_TR_2 first, then M_R

remaining_vars2 = ["M_R", "R"]

# after M_TR_2 cut first
xc_mtr_first, _ = find_best_cut(df_sig, df_bkg, "M_TR_2", cut_types["M_TR_2"], 1000, 10000)
if cut_types["M_TR_2"] == "x > x_c":
    df_sig_rev1 = df_sig[df_sig["M_TR_2"] > xc_mtr_first]
    df_bkg_rev1 = df_bkg[df_bkg["M_TR_2"] > xc_mtr_first]
else:
    df_sig_rev1 = df_sig[df_sig["M_TR_2"] < xc_mtr_first]
    df_bkg_rev1 = df_bkg[df_bkg["M_TR_2"] < xc_mtr_first]

print("After M_TR_2 cut first:")
plot_roc(df_sig_rev1, df_bkg_rev1, remaining_vars2, cut_types, "ROC after M_TR_2 cut (reversed order)")

print("If the variables are correlated, the order of cuts can change the ROC curves.")
print("Less correlated variables are less affected by the order.")
```

## Exercise 8: Linear Discriminant

```python
# Exercise 8.1 - compute S_B and S_W

# use all observables (skip signal column)
obs_names = VarNames[1:]

# get signal and background as numpy arrays
X_sig = df_sig[obs_names].values
X_bkg = df_bkg[obs_names].values

# compute means
m1 = np.mean(X_sig, axis=0)  # signal mean vector
m2 = np.mean(X_bkg, axis=0)  # background mean vector

# between-class covariance S_B = (m2 - m1)(m2 - m1)^T
diff = (m2 - m1).reshape(-1, 1)  # make it a column vector
S_B = diff @ diff.T

# within-class covariance S_W
# S_W = sum over signal of (x - m1)(x - m1)^T + sum over background of (x - m2)(x - m2)^T
# this is the same as (n1-1)*cov_sig + (n2-1)*cov_bkg

cov_sig = np.cov(X_sig, rowvar=False)
cov_bkg = np.cov(X_bkg, rowvar=False)

S_W = (len(X_sig) - 1) * cov_sig + (len(X_bkg) - 1) * cov_bkg

print("S_B shape:", S_B.shape)
print("S_W shape:", S_W.shape)
```

```python
# Exercise 8.2 - compute linear coefficients w = S_W^-1 (m2 - m1)
w = np.linalg.inv(S_W) @ (m2 - m1)

print("Linear discriminant weights:")
for i, name in enumerate(obs_names):
    print("  ", name, ":", round(w[i], 6))
```

```python
# compute the discriminant F = w^T * x for signal and background
F_sig = X_sig @ w
F_bkg = X_bkg @ w

# histogram the discriminant
plt.figure(figsize=(10, 5))
plt.hist(F_sig, bins=100, histtype="step", color="red", label="signal", density=1)
plt.hist(F_bkg, bins=100, histtype="step", color="blue", label="background", density=1)
plt.xlabel("F (linear discriminant)")
plt.ylabel("normalized")
plt.title("Linear Discriminant Distribution")
plt.legend()
plt.show()
```

```python
# Exercise 8.3 - ROC curve for the linear discriminant
# since F_sig should be different from F_bkg, we can cut on F

# try both directions and pick the better one
cut_values_F = np.linspace(np.percentile(np.concatenate([F_sig, F_bkg]), 1),
                            np.percentile(np.concatenate([F_sig, F_bkg]), 99), 200)

# compute TPR/FPR for F > x_c
tpr_list = list()
fpr_list = list()
n_sig = len(F_sig)
n_bkg = len(F_bkg)

for xc in cut_values_F:
    tpr_list.append(np.sum(F_sig > xc) / float(n_sig))
    fpr_list.append(np.sum(F_bkg > xc) / float(n_bkg))

tpr_F = np.array(tpr_list)
fpr_F = np.array(fpr_list)

auc_F = compute_auc(fpr_F, tpr_F)

# if AUC < 0.5 we need to flip the direction
if auc_F < 0.5:
    tpr_F = 1 - tpr_F
    fpr_F = 1 - fpr_F
    auc_F = compute_auc(fpr_F, tpr_F)

plt.figure(figsize=(8, 8))
plt.plot(fpr_F, tpr_F, label="Linear Discriminant (AUC=" + str(round(auc_F, 3)) + ")")

# overlay the best individual variables for comparison
for var in ["M_R", "M_TR_2", "R"]:
    fpr_v, tpr_v = compute_roc(df_sig, df_bkg, var, cut_types[var])
    auc_v = compute_auc(fpr_v, tpr_v)
    plt.plot(fpr_v, tpr_v, '--', label=var + " (AUC=" + str(round(auc_v, 3)) + ")")

plt.plot([0, 1], [0, 1], 'k--', label="random")
plt.xlabel("FPR")
plt.ylabel("TPR")
plt.title("ROC: Linear Discriminant vs Individual Variables")
plt.legend()
plt.show()
```

```python
# Exercise 8.4 - maximal significance using the linear discriminant for each scenario
print("Max significance using linear discriminant:")
print()

for N_S, N_B in scenarios:
    best_sig = 0
    for xc in cut_values_F:
        eps_s = np.sum(F_sig > xc) / float(n_sig)
        eps_b = np.sum(F_bkg > xc) / float(n_bkg)
        N_S_prime = eps_s * N_S
        N_B_prime = eps_b * N_B
        if N_S_prime + N_B_prime > 0:
            significance = N_S_prime / np.sqrt(N_S_prime + N_B_prime)
        else:
            significance = 0
        if significance > best_sig:
            best_sig = significance
    
    print("N_S =", N_S, "N_B =", N_B, ": max significance =", round(best_sig, 3))
```
